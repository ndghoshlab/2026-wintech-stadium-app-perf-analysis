# B-Browsing analysis

### plc, data creation

```python
df_browser = pd.concat([
    pd.read_csv(DATA_DIR / 'game5_qp_csv/20251108_Game_Bowl_Browser.csv').assign(Location='Bowl'),
    pd.read_csv(DATA_DIR / 'game5_qp_csv/20251108_Game_Concourse_Browser.csv').assign(Location='Concourse')
])

df_browser['Time'] = pd.to_datetime(df_browser['Time'])
df_browser = df_browser.sort_values('Time')
display(df_browser[['Unit', 'Operator']].value_counts())

# Taken from SigCap
uuid_map = {
    'QualiPoc_619529': '64957d65',	 #T-Mobile
    'QualiPoc_361786': 'a0654bd3',	 #AT&T
    'QualiPoc_396840': '7c36e7df'	 #Verizon
}

df_browser['uuid'] = df_browser['Unit'].map(uuid_map)

print(f"Browser data shape: {df_browser.shape}")
df_browser.head()
```

```python
print("Original shape:", df_browser.shape)
```

```python
# 1. Identify columns
group_cols = [col for col in df_browser.columns if col != "Data Technology Reporting"]

# 2. Group with dropna=False
df_collapsed = (
    df_browser.fillna({"Data Technology Reporting": "NaN"})
    .groupby(group_cols, as_index=False, dropna=False)["Data Technology Reporting"]
    .agg(", ".join)
)

print("New shape:", df_collapsed.shape)
df_browser = df_collapsed.copy()
```

```python
df_browser[df_browser['Location'] == 'Bowl'].describe()
```

```python
report_lines = []

df_operate = df_browser.copy()

df_operate = df_operate[
    ~(df_operate['Time'] == '')
    & (df_operate['Operator'] == 'AT&T')
    & (df_operate['Location'] == 'Bowl')
    # & (df_dl_ca_browsing['String_Band'].str.contains('b2', na=False))
].copy()

df_operate = df_operate[
    ~(df_operate['Time'] == '')
    # & (
    #     (df_operate['Browsing Duration'] >= df_operate['Browsing Duration'].quantile(0.75))
    #     | (df_operate['Time To First Byte'] >= df_operate['Time To First Byte'].quantile(0.75))
    # )
    # & (df_operate['Time To First Byte'] ==  11581)
    
].copy()

ca_counts = df_operate[[
    # 'timeslot_interval', 
    # 'Time To First Byte',
    # 'Host', 
    # 'Size Of Webpage', 
    # 'Operator_QP', 
    'Time', 
    # 'Time', 'Timestamp Metric',
    'Latitude', 'Longitude', 
    # 'Throughput', 'IP Throughput',
    # 'String_Band', 'String_ARFCN',
    # 'Technology',
    'Data Technology Reporting',
    'Browsing Duration',
    # 'Size Of Webpage', 'Host'
]].value_counts(dropna=False).sort_index().to_string()

report_lines.append(ca_counts)


final_report = "\n".join(report_lines)
print(final_report)

# Save to the text file
# (OUTPUT_DIR / 'ATT_slower-than-median.txt').write_text(final_report, encoding='utf-8')
```

```python
# Correlate df_browser and df_sigcap

# 1. Handle Timezones
# Browser 'Time' is likely naive local time (ET). Localize it.
if df_browser['Time'].dt.tz is None:
    df_browser['Time'] = df_browser['Time'].dt.tz_localize('US/Eastern', ambiguous='infer')

# Sigcap 'timestamp' is already tz-aware. Convert both to UTC for safe merging.
df_browser['Time_utc'] = df_browser['Time'].dt.tz_convert('UTC')
df_sigcap['timestamp_utc'] = df_sigcap['timestamp'].dt.tz_convert('UTC')

# 2. Merge
# We want to bring 'network_type*' and 'wifi_connected_primary_freq_mhz' from sigcap to browser
cols_to_merge = ['timestamp_utc', 'operator', 'network_type*', 'wifi_connected_primary_freq_mhz', 'Location']

cols_to_merge.append('uuid')
df_browser_merged = pd.merge_asof(
    df_browser,
    df_sigcap[cols_to_merge],
    left_on='Time_utc',
    right_on='timestamp_utc',
    by='uuid',
    direction='nearest',
    tolerance=pd.Timedelta('10s')
)

# 3. Cleanup
df_browser_merged = df_browser_merged.drop(columns=['timestamp_utc', 'Time_utc'])

# Rename Location columns for clarity after merge
df_browser_merged = df_browser_merged.rename(columns={'Location_x': 'Location', 'Location_y': 'Location_sigcap'})

conditions = [
    (df_browser_merged['wifi_connected_primary_freq_mhz'].between(5000, 5925, inclusive='both')),
    (df_browser_merged['wifi_connected_primary_freq_mhz'] > 5925),
    (df_browser_merged['wifi_connected_primary_freq_mhz'] < 5000)
]
choices = ['Wi-Fi 5', 'Wi-Fi 6', 'Wi-Fi 2.4']
df_browser_merged['operator'] = np.select(conditions, choices, default=df_browser_merged['operator'])

# Check for Location mismatches
df_mismatched_Locations = df_browser_merged[df_browser_merged['Location'] != df_browser_merged['Location_sigcap']]
print(f"Rows with mismatched Locations: {df_mismatched_Locations.shape[0]}")
display(df_mismatched_Locations)

# Display result
print(f"Merged shape: {df_browser_merged.shape}")
display(df_browser_merged)
df_browser_merged.to_csv(OUTPUT_DIR / 'df_browser_merged_gd5.csv', index=False)
```
