# B- RAN Configuration Distribution

**This one is USED in the paper**

```python
import pandas as pd

# 1. Prepare data
df_look_ran = look_radio.copy()
df_look_ran = df_look_ran[df_look_ran['RAN Configuration'] != 'WiFi'].copy()

# 2. Deduplicate safely by taking the mode of 'RAN Configuration'
df_deduped = (
    df_look_ran.groupby(['Time', 'Label', 'Operator'])['RAN Configuration']
    .agg(lambda x: m[0] if len(m := x.mode()) > 0 else None)
    .reset_index()
)

# 3. Get value counts from the deduplicated data
counts = (
    df_deduped[['Label', 'Operator', 'RAN Configuration']]
    .value_counts(dropna=True)
    .sort_index()
)

# 4. Calculate percentages within each ('Label', 'Operator') group
percentages = counts.groupby(level=['Label', 'Operator']).transform(
    lambda x: round(x / x.sum() * 100, 2)
)

# 5. Combine into final dataframe and display
result = pd.DataFrame({'Count': counts, 'Percentage (%)': percentages})
print(result)
```
